```python
import pandas as pd
train1 = pd.read_csv('Fake.csv')
train1.head()
```

```python
train2 = pd.read_csv('True.csv')
train2.head()
```

```python
train1['label']=0
train1.head()
```

```python
train2['label']=1
train2.head()
```

```python
frames = [train1, train2]
train = pd.concat(frames)
```

```python
train.head(-15)
```

```python
# Firstly, fill all the null spaces with a space
train = train.fillna(' ')
train['total']=train['title']+' '+train['text']+' '+train['subject']+' '+train['date']
```

```python
train.head(-1)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
#nltk.download('stopwords')
```

```python
stop_words = stopwords.words('english')
```

```python
lemmatizer = WordNetLemmatizer()
import re
#nltk.download('punkt')
```

```python
for index, row in train.iterrows():
    filter_sentence = ''
    sentence = row['total']
    # Cleaning the sentence with regex
    sentence = re.sub(r'[^\w\s]', '', sentence)
    # Tokenization
    words = nltk.word_tokenize(sentence,language="english")
    # Stopwords removal
    words = [w for w in words if not w in stop_words]
    # Lemmatization
    for words in words:
        filter_sentence = filter_sentence  + ' ' +str(lemmatizer.lemmatize(words)).lower()
        train.loc[index, 'total'] = filter_sentence
```
